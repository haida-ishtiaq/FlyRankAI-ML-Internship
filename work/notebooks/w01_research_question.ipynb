## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The question: Which pages should a content team review first for refresh, given limited time?

The decision this improves: Which pages get a human content reviewer's attention this week/month, out of thousands of possible pages.

The action someone takes: A content strategist opens the top-ranked pages first and decides whether to update the content, expand it, fix metadata, or leave it alone.

The cost of a wrong call:

If the model flags a healthy page as "needs refresh" (false positive), the team wastes reviewer time on a page that didn't need it — a small time/opportunity cost.
If the model misses a genuinely declining page (false negative), that page keeps losing visibility/traffic silently, which is a bigger cost because it compounds over time and the team never even looks at it.

This asymmetry is why a ranked list is safer than a single yes/no cutoff — it lets a human use judgment.
